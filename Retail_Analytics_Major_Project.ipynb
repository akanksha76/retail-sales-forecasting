```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Use the daily series created in Step 16.
# Fill missing dates with zero only if they genuinely represent
# days with no sales.
daily_revenue = observed_daily.reindex(
    all_dates,
    fill_value=0
).astype(float)

# Split chronologically: last 60 days for testing
test_size = 60

train = daily_revenue.iloc[:-test_size]
test = daily_revenue.iloc[-test_size:]

print("Training observations:", len(train))
print("Testing observations:", len(test))

# Model 1: Seasonal-naive baseline (weekly pattern)
baseline_predictions = pd.Series(
    np.resize(train.iloc[-7:].to_numpy(), len(test)),
    index=test.index
)

# Model 2: Holt-Winters with weekly seasonality
hw_model = ExponentialSmoothing(
    train,
    trend="add",
    seasonal="add",
    seasonal_periods=7,
    initialization_method="estimated"
).fit()

hw_predictions = pd.Series(
    hw_model.forecast(len(test)).to_numpy(),
    index=test.index
)

# Evaluate both models
results = []

for name, predictions in [
    ("Seasonal Naive", baseline_predictions),
    ("Holt-Winters", hw_predictions)
]:
    mae = mean_absolute_error(test, predictions)
    rmse = np.sqrt(mean_squared_error(test, predictions))

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse
    })

metrics_df = pd.DataFrame(results).sort_values("RMSE")

print("\nForecasting Model Comparison:")
display(metrics_df.round(2))

# Plot actual vs predicted revenue over the test period
plt.figure(figsize=(14, 6))

plt.plot(
    test.index,
    test.values,
    label="Actual Revenue"
)

plt.plot(
    test.index,
    baseline_predictions.values,
    label="Seasonal Naive"
)

plt.plot(
    test.index,
    hw_predictions.values,
    label="Holt-Winters"
)

plt.title("Actual vs Predicted Daily Revenue — Test Period")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Convert revenue to millions for model fitting
scale = 1_000_000
series = daily_revenue.astype(float)

# Keep the latest 60 days for testing
train_scaled = series.iloc[:-60] / scale
test_actual = series.iloc[-60:]
test_scaled = test_actual / scale

# Baseline: repeat the last observed 7-day pattern
baseline_pred = pd.Series(
    np.resize(train_scaled.iloc[-7:].to_numpy(), len(test_actual))
    * scale,
    index=test_actual.index
)

# Fit Holt-Winters on scaled training data
evaluation_model = ExponentialSmoothing(
    train_scaled,
    trend="add",
    seasonal="add",
    seasonal_periods=7,
    initialization_method="estimated"
).fit(optimized=True, use_brute=False)

hw_pred = pd.Series(
    evaluation_model.forecast(len(test_actual)).to_numpy() * scale,
    index=test_actual.index
)

# Compare models
results = []

for name, pred in [
    ("Seasonal Naive", baseline_pred),
    ("Holt-Winters", hw_pred)
]:
    results.append({
        "Model": name,
        "MAE": mean_absolute_error(test_actual, pred),
        "RMSE": np.sqrt(mean_squared_error(test_actual, pred)),
        "WAPE (%)": (
            np.abs(test_actual - pred).sum()
            / np.abs(test_actual).sum()
        ) * 100
    })

metrics_df = pd.DataFrame(results).sort_values("RMSE")
display(metrics_df.round(2))

# Visualize the test predictions
plt.figure(figsize=(14, 5))
plt.plot(test_actual.index, test_actual.values, label="Actual Revenue")
plt.plot(hw_pred.index, hw_pred.values, label="Holt-Winters Prediction")
plt.title("Holt-Winters Model Evaluation — Last 60 Days")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Prepare the daily revenue series
series = daily_revenue.sort_index().astype(float)

# Reserve the last 60 days for testing
test_size = 60
split_idx = len(series) - test_size

train = series.iloc[:split_idx]
test = series.iloc[split_idx:]

print("Training days:", len(train))
print("Testing days:", len(test))


# Create features using only past observations
def create_features(history, date):
    history = np.asarray(history, dtype=float)

    return {
        "lag_1": history[-1],
        "lag_7": history[-7],
        "lag_14": history[-14],
        "lag_28": history[-28],
        "mean_7": history[-7:].mean(),
        "mean_28": history[-28:].mean(),
        "std_7": history[-7:].std(),
        "day_of_week": date.dayofweek,
        "day_of_month": date.day,
        "month": date.month,
        "day_of_year": date.dayofyear,
        "is_weekend": int(date.dayofweek >= 5)
    }


feature_names = [
    "lag_1", "lag_7", "lag_14", "lag_28",
    "mean_7", "mean_28", "std_7",
    "day_of_week", "day_of_month", "month",
    "day_of_year", "is_weekend"
]

# Build training examples without using future values
X_train = []
y_train = []

for i in range(28, split_idx):
    history = series.iloc[:i].to_numpy()
    date = series.index[i]

    X_train.append(create_features(history, date))
    y_train.append(series.iloc[i])

X_train = pd.DataFrame(X_train, columns=feature_names)
y_train = np.asarray(y_train)

print("Random Forest training samples:", len(X_train))


# Train the model
rf_model = RandomForestRegressor(
    n_estimators=300,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)


# Generate recursive predictions for the test period
history = list(train.to_numpy())
rf_predictions = []

for date in test.index:
    features = create_features(history, date)
    X_next = pd.DataFrame([features], columns=feature_names)

    prediction = rf_model.predict(X_next)[0]
    prediction = max(float(prediction), 0)

    rf_predictions.append(prediction)

    # Use the prediction for subsequent forecast features
    history.append(prediction)

rf_predictions = pd.Series(
    rf_predictions,
    index=test.index
)

print("Random Forest forecasting completed!")
```

```python
# Seasonal-naive baseline
baseline_predictions = pd.Series(
    np.resize(train.iloc[-7:].to_numpy(), len(test)),
    index=test.index
)

# Holt-Winters trained only on the training period
scale = 1_000_000

hw_eval_model = ExponentialSmoothing(
    train / scale,
    trend="add",
    seasonal="add",
    seasonal_periods=7,
    initialization_method="estimated"
).fit(optimized=True, use_brute=False)

hw_predictions = pd.Series(
    np.maximum(
        hw_eval_model.forecast(len(test)).to_numpy() * scale,
        0
    ),
    index=test.index
)


# Evaluate all models
model_predictions = {
    "Seasonal Naive": baseline_predictions,
    "Holt-Winters": hw_predictions,
    "Random Forest": rf_predictions
}

comparison = []

for name, pred in model_predictions.items():
    comparison.append({
        "Model": name,
        "MAE": mean_absolute_error(test, pred),
        "RMSE": np.sqrt(mean_squared_error(test, pred)),
        "WAPE (%)": (
            np.abs(test - pred).sum()
            / np.abs(test).sum()
        ) * 100
    })

comparison_df = (
    pd.DataFrame(comparison)
    .sort_values("RMSE")
    .reset_index(drop=True)
)

display(comparison_df.round(2))


# Plot actual vs predicted revenue
plt.figure(figsize=(14, 6))

plt.plot(test.index, test.values, label="Actual Revenue")
plt.plot(
    baseline_predictions.index,
    baseline_predictions.values,
    label="Seasonal Naive"
)
plt.plot(
    hw_predictions.index,
    hw_predictions.values,
    label="Holt-Winters"
)
plt.plot(
    rf_predictions.index,
    rf_predictions.values,
    label="Random Forest"
)

plt.title("Forecasting Model Comparison — Test Period")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# Save the comparison table
comparison_df.to_csv(
    "/content/forecast_model_comparison.csv",
    index=False
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Prepare the series and split chronologically
series = daily_revenue.sort_index().astype(float)

test_size = 60
split_idx = len(series) - test_size

train = series.iloc[:split_idx]
test = series.iloc[split_idx:]

print("Training days:", len(train))
print("Testing days:", len(test))

# 2. Seasonal Naive baseline
baseline_predictions = pd.Series(
    np.resize(train.iloc[-7:].to_numpy(), len(test)),
    index=test.index
)

# 3. Holt-Winters model
scale = 1_000_000

hw_model = ExponentialSmoothing(
    train / scale,
    trend="add",
    seasonal="add",
    seasonal_periods=7,
    initialization_method="estimated"
).fit(optimized=True, use_brute=False)

hw_predictions = pd.Series(
    np.maximum(hw_model.forecast(len(test)).to_numpy() * scale, 0),
    index=test.index
)

# 4. Random Forest features
def create_features(history, date):
    history = np.asarray(history, dtype=float)
    return {
        "lag_1": history[-1],
        "lag_7": history[-7],
        "lag_14": history[-14],
        "lag_28": history[-28],
        "mean_7": history[-7:].mean(),
        "mean_28": history[-28:].mean(),
        "std_7": history[-7:].std(),
        "day_of_week": date.dayofweek,
        "day_of_month": date.day,
        "month": date.month,
        "day_of_year": date.dayofyear,
        "is_weekend": int(date.dayofweek >= 5)
    }

feature_names = [
    "lag_1", "lag_7", "lag_14", "lag_28",
    "mean_7", "mean_28", "std_7",
    "day_of_week", "day_of_month", "month",
    "day_of_year", "is_weekend"
]

X_train, y_train = [], []

for i in range(28, split_idx):
    history = series.iloc[:i].to_numpy()
    X_train.append(create_features(history, series.index[i]))
    y_train.append(series.iloc[i])

X_train = pd.DataFrame(X_train, columns=feature_names)

rf_model = RandomForestRegressor(
    n_estimators=300,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, np.asarray(y_train))

# Predict test period recursively
history = list(train.to_numpy())
rf_values = []

for date in test.index:
    features = create_features(history, date)
    X_next = pd.DataFrame([features], columns=feature_names)
    prediction = max(float(rf_model.predict(X_next)[0]), 0)
    rf_values.append(prediction)
    history.append(prediction)

rf_predictions = pd.Series(rf_values, index=test.index)

# 5. Compare model performance
model_predictions = {
    "Seasonal Naive": baseline_predictions,
    "Holt-Winters": hw_predictions,
    "Random Forest": rf_predictions
}

results = []

for name, pred in model_predictions.items():
    results.append({
        "Model": name,
        "MAE": mean_absolute_error(test, pred),
        "RMSE": np.sqrt(mean_squared_error(test, pred)),
        "WAPE (%)": (
            np.abs(test - pred).sum() / np.abs(test).sum()
        ) * 100
    })

comparison_df = (
    pd.DataFrame(results)
    .sort_values("RMSE")
    .reset_index(drop=True)
)

display(comparison_df.round(2))

# 6. Plot actual vs predicted revenue
plt.figure(figsize=(14, 6))
plt.plot(test.index, test.values, label="Actual Revenue")
plt.plot(baseline_predictions.index, baseline_predictions.values,
         label="Seasonal Naive")
plt.plot(hw_predictions.index, hw_predictions.values,
         label="Holt-Winters")
plt.plot(rf_predictions.index, rf_predictions.values,
         label="Random Forest")

plt.title("Actual vs Predicted Daily Revenue — Test Period")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# 7. Save comparison results
comparison_df.to_csv(
    "/content/forecast_model_comparison.csv",
    index=False
)

print("Model comparison saved successfully!")
```
